```python
import yfinance as yf
from datetime import date, timedelta
# import pandas as pd 

fetch_start = "2020-01-01"  # first run fallback

crude = yf.download("^DJI", start=fetch_start, progress=True)
```

```python
crude.columns = crude.columns.droplevel("Ticker")
```

```python
df_crude = crude.reset_index().rename(columns={
    "Date": "date", 
    "Open": "open", 
    "High": "high", 
    "Low": "low", 
    "Close": "close", 
    "Volume": "volume"})
```

```python
import yfinance as yf
```

```python
ticker="NVDA"
start_date="2020-01-01"

stock_price = yf.download(ticker, start=start_date, progress=True)
```

```python
## drop Ticker row header
stock_price.columns=stock_price.columns.droplevel("Ticker")
```

```python
## reset index to keep just one header
stock_price = stock_price.reset_index()
```

```python
stock_price.columns
```

```python
import pandas as pd 

stock_price["Date"] = pd.to_datetime(stock_price["Date"])
```

```python
import matplotlib.pyplot as plt
plt.style.use("fivethirtyeight")
plt.figure(figsize=(10, 6))
stock_price.plot(x="Date", y="Close")
plt.title(f"{ticker} Closing Price")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.grid(True)
plt.show()
```
